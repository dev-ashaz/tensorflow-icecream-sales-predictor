```python
import tensorflow as tf
import numpy as np

print("ICE CREAM SALES PREDICTOR")

temperature = np.array(
    [15, 18, 20, 22, 25, 28, 30, 32],
    dtype=float
)

sales = np.array(
    [40, 55, 70, 85, 110, 140, 160, 180],
    dtype=float
)

model = tf.keras.Sequential([
    tf.keras.layers.Dense(
        units=1,
        input_shape=[1]
    )
])

model.compile(
    optimizer='sgd',
    loss='mean_squared_error'
)

model.fit(
    temperature,
    sales,
    epochs=500,
    verbose=0
)

prediction = model.predict(np.array([[35.0]]))

print(
    f"Predicted Ice Cream Sales at 35°C: "
    f"{prediction[0][0]:.2f}"
)
```
